```python
# We have something called SONAR used in submarines,
# where the data from that SONAR is used to tell if the object detected by the SONAR is a rock or a mine
# kinda military use case.... 
# work flow : The Sonar data from lab is collected -> Then the data is undergone some preprocessing
# -> now we will split our data to train data and test data -> this data is feed into a model(for this case we will use Logistic Regression model)
# -> then we will take this trained model and take some real world or new data that the model is not trained with and we check how accurate it is...
```

```python
# since everything is in numaric value, no need to do vectorization
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
# Now we work with the data
datadf = pd.read_csv("Copy of sonar data.csv", header=None)
# our dataset has no headers, so have to let the pandas know that there are no headers,
# if we dont the first row(in our case which is actually data not header of table), pandas will take that as header
# now since we told panda that we don't have any header it will put indexes
```

```python
datadf.head()
# the indexes start from 0 to n-1
```

```python
# now the number of rows and columns are
print(datadf.shape)
# we have 208 rows and 60 features + 1 target(result/label)
```

```python
# now lets see how man rock and how many mines we have in our data set
datadf[60].value_counts()
# output:
# M    111
# R     97
# Name: count, dtype: int64
# the number of M and R are almost equal so, its fine.....
# the data is less but for a small model, its alright
# but if we get more data, more accurate our model will be
```

```python
# Now we can check the mean value of the values that are mine or rock
# we can use groupby() to group these M and R and then mean() to get the mean
datadf.groupby(60).mean()
```

```python
# Now we are gonna separate data and label
X= datadf.drop(columns=60, axis=1) # 0-59 columns
Y=datadf[60] # 60th column
```

```python
# Now we will split the data based on train and test
X_train,X_test,Y_train, Y_test = train_test_split(X,Y, test_size=0.15, stratify=Y, random_state=4)
```

```python
model = LogisticRegression()
# the model: LogisticRegression
```

```python
model.fit(X_train, Y_train)
# train the model with training data
```

```python
# Now we can check the accuracy how well the model is trained
# ofc since the data is less, the accuracy will be low
# now we will trian to test with the test data, that model have never seen
X_test_prediction = model.predict(X_test)
# now the model predicts the data of Y_test using the X_test
# now that we will compare with the Y_test
acc_score = accuracy_score(X_test_prediction, Y_test)
print(acc_score)
#ouptut: 0.75
# that means out of 100 values 75 will be right, might look bad but the accuracy it has got just from such small data is awsome
# we only had like 200 data sets, in real we get 100000+++, it will be way more accurate than this
```
